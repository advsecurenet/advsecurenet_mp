# GPT-2 Fine-tuning with AdvSecureNet

Simple example demonstrating GPT-2 fine-tuning on WikiText-2 dataset using only the enhanced components from `llm_finetuning`.

```python
# Setup and imports
import sys
import warnings
warnings.filterwarnings('ignore')

# Add project path and import our enhanced system
sys.path.append('/Users/philip/Desktop/advsecurenet_mp')
from advsecurenet.llm_finetuning import (
    ConfigManager, ModelConfig, DataConfig, TrainingConfig, LanguageModelingPipeline
)

# Load small dataset from HuggingFace
from datasets import load_dataset
dataset = load_dataset("wikitext", "wikitext-2-raw-v1")

# Prepare small subset for demo
train_texts = [text for text in dataset['train']['text'][:500] if len(text.strip()) > 20][:200]
val_texts = [text for text in dataset['validation']['text'][:100] if len(text.strip()) > 20][:50]

print(f"Training samples: {len(train_texts)}")
print(f"Validation samples: {len(val_texts)}")
```

## Configuration Setup

```python
# Create configuration using our enhanced system
config_manager = ConfigManager()

# Model configuration
config_manager.model_config = ModelConfig(
    model_name="gpt2",
    model_type="language_modeling"
)

# Data configuration
config_manager.data_config = DataConfig(
    text_column="text",
    max_length=128,
    preprocessing_num_workers=2,
    validation_split=0.2
)

# Training configuration
config_manager.training_config = TrainingConfig(
    output_dir="./results/simple_gpt2",
    num_train_epochs=2,
    per_device_train_batch_size=4,
    learning_rate=5e-5,
    logging_steps=25,
    eval_steps=50,
    save_steps=100,
    evaluation_strategy="steps"
)

# Save configuration
config_path = "./simple_gpt2_config.yaml"
config_manager.save_config(config_path)
print(f"Configuration saved: {config_path}")
```

## Training

```python
# Initialize and run training pipeline
pipeline = LanguageModelingPipeline(config_path)

print("Starting training...")
results, trained_model, trained_tokenizer = pipeline.run_from_texts(
    train_texts=train_texts,
    val_texts=val_texts,
    early_stopping=True
)

print("Training completed!")
```

## Text Generation

```python
# Generate text using trained model
from advsecurenet.llm_finetuning import Trainer

trainer = Trainer(trained_model, trained_tokenizer, config_manager, None)

# Generate sample text
prompt = "The development of artificial intelligence"
generated = trainer.generate_text(
    prompt=prompt,
    max_length=100,
    temperature=0.7,
    num_return_sequences=2
)

print(f"Prompt: {prompt}")
for i, text in enumerate(generated, 1):
    print(f"\nGeneration {i}:")
    print(text)
```

## Results

Model successfully fine-tuned and capable of generating domain-specific text. All components used from the enhanced `llm_finetuning` system.
