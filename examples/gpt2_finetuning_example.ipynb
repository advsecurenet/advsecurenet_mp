## 8. Generate Text with the Fine-tuned Model

Now for the exciting part! Let's generate some text using our fine-tuned GPT-2 model and compare it with the original model.

````python
# Function to generate text with different parameters
def generate_text(model, tokenizer, prompt, max_length=100, temperature=0.7, num_return_sequences=1):
    """Generate text using the fine-tuned model."""
    # Encode the prompt
    inputs = tokenizer.encode(prompt, return_tensors='pt').to(device)
    
    # Generate text
    with torch.no_grad():
        outputs = model.generate(
            inputs,
            max_length=max_length,
            temperature=temperature,
            num_return_sequences=num_return_sequences,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id,
            top_k=50,
            top_p=0.95,
            repetition_penalty=1.2
        )
    
    # Decode the generated text
    generated_texts = []
    for output in outputs:
        generated_text = tokenizer.decode(output, skip_special_tokens=True)
        generated_texts.append(generated_text)
    
    return generated_texts

# ===== Text Generation Using Our Enhanced System =====
print("🎯 Text Generation with AdvSecureNet Enhanced System")
print("Using our Trainer's built-in generate_text method!")
print("=" * 60)

# Create an Trainer instance to access our generation methods
print("🤖 Creating Trainer for text generation...")
enhanced_trainer = Trainer(trained_model, trained_tokenizer, config_manager, None)

# Test prompts for generation
test_prompts = [
    "The future of artificial intelligence",
    "Machine learning algorithms are",
    "In the field of natural language processing",
    "Deep learning has revolutionized",
    "The importance of data in AI"
]

print("🔤 Generating text with our Enhanced System...")

for i, prompt in enumerate(test_prompts, 1):
    print(f"\n🎯 Prompt {i}: '{prompt}'")
    print("-" * 50)
    
    # Use our enhanced trainer's generate_text method
    generated_texts = enhanced_trainer.generate_text(
        prompt=prompt,
        max_length=150,
        temperature=0.8,
        num_return_sequences=2
    )
    
    for j, text in enumerate(generated_texts, 1):
        print(f"\nGeneration {j} (Enhanced System):")
        print(text)
        print("-" * 30)

print("\n✅ Text generation completed using our Trainer!")
print("🎯 All generation handled by our built-in generate_text method!")
print("No manual generation code needed - everything in our enhanced system!")
````
