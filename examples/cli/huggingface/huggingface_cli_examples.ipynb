## Step 2: Training Examples

Following the API examples, here are different approaches for HuggingFace training configuration:

1. **Default approach** (`huggingface_train_config.yml`): Uses `load_splits` to load both train and test splits
2. **Split config approach** (`huggingface_train_split_config.yml`): Uses `split_config` to explicitly define train/test splits with different identifiers

Note: Both approaches achieve the same result of loading HuggingFace datasets, but `split_config` provides more flexibility for complex dataset configurations.

```python
# Generate default HuggingFace training config
!advsecurenet utils configs get -c huggingface_train_config.yml -s -o ./hf_train_default.yml
```

```python
# Generate training config with load_splits approach (default HuggingFace config uses load_splits)
!advsecurenet utils configs get -c huggingface_train_load_splits_config.yml -s -o ./hf_train_load_splits.yml
```

```python
# Generate split_config approach training config
!advsecurenet utils configs get -c huggingface_train_split_config.yml -s -o ./hf_train_split_config.yml
```
