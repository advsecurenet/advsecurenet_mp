# Distributed DPatch Attack
Adversarial patch optimisation is heavier than single-step pixel attacks. This notebook shows how to configure and launch the CLI-driven **DPatch** object-detection attack in distributed mode so every GPU contributes patch gradients while staying in sync.

## Using a Configuration File


### Step 1: Create a Configuration File

Run the helper command below from the repository root (`advsecurenet_mp`) to materialise a copy you can edit without touching the shipped defaults.

````python
!advsecurenet utils configs get -c dpatch_attack_config.yml -s -o ./default_dpatch_attack_config.yml
````

## Step 2: Modify the Configuration File

Key edits before launching the distributed run:
- Set `device.use_ddp: True` so the CLI spawns one worker per GPU through the internal `DDPCoordinator`.
- Point `device.gpu_ids` to the target CUDA devices you want to run the attack on.

We will set `device.use_ddp` to *True* and define `device_gpu_ids` as a list: [5, 6, 7]. We will also increase the `random_sample_size` to 90 and decrease the `batch_size` to 5. You can find the updated configuration file [here](dpatch_attack_config.yml).

## Step 3: Run the Attack

With the YAML saved (for example as `dpatch_attack_config.yml`), launch the attack from the repository root. The CLI reads the DDP flags, spawns one process per GPU, broadcasts the initial patch, and synchronises gradients every optimisation step.

```sh
advsecurenet attack dpatch -c dpatch_attack_config.yml
```

**Note:** Attack success is reported via the change in mean average precision before and after applying the trained patch. Because the example config uses a small evaluation subset, the numbers may differ from full-dataset expectations.

````python
!advsecurenet attack dpatch -c dpatch_attack_config.yml
````
