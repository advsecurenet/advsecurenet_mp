# Distributed TOG Attack
This notebook shows how to configure and launch the CLI-driven **TOG** object-detection attack in distributed mode so every GPU contributes pixel perturbations while staying in sync

## Using a Configuration File


### Step 1: Create a Configuration File

As a first step, you need to have a default configuration file. You can create a configuration file by running the following command (from the repository root folder `advsecurenet_mp`):
    
```sh
advsecurenet utils configs get -c tog_attack_config.yml -s -o ./default_tog_attack_config.yml
```

Here, we are also providing the `-o` flag to specify the output directory where the configuration file will be saved. You can also specify the output directory by passing the `-o` flag followed by the path to the directory where you want to save the configuration file. If you don't provide the `-o` flag, the configuration file will be saved in the current working directory with the name `tog_attack_config.yml`. If the file already exists, it will be overwritten. If the output path is a directory, the configuration file will be saved in that directory with the name `tog_attack_config.yml`.

## Step 2: Modify the Configuration File

Key edits before launching the distributed run:
- Set `device.use_ddp: True` so the CLI spawns one worker per GPU through the internal `DDPCoordinator`.
- Point `device.gpu_ids` to the target CUDA devices you want to run the attack on.

You can find the updated configuration file [here](tog_attack_config.yml).

## Step 3: Run the Attack

Now that you have created and modified the configuration file, you can run the attack in DDP mode by using the following command:

```sh
advsecurenet attack tog -c tog_attack_config.yml
```

**Note:** Attack success rate is calculated as difference in the mean average precision which is the average precision aggregated over all of the classes within the dataset before and after the attack. Since the attack runs on a small subset of the whole dataset, which might not contain enough representatives of each class, the attack success rate might be lower than expected.

````python
!advsecurenet attack tog -c tog_attack_config.yml
````
