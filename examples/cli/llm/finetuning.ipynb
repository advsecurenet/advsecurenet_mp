# LLM Fine-tuning CLI Examples

This notebook demonstrates how to use the AdvSecureNet LLM fine-tuning CLI for training language models with various configurations and techniques.

## Overview

The fine-tuning CLI supports:
- **PEFT (Parameter-Efficient Fine-Tuning)** with LoRA
- **Full fine-tuning** for complete model adaptation
- **Instruction tuning** for chat/assistant models
- **Custom dataset integration**
- **Multi-GPU training support**

## Prerequisites

Make sure you have:
- A YAML configuration file (see examples in `configs/`)
- Training data in the correct format (JSON recommended)
- Sufficient GPU memory or CPU resources
- Required dependencies installed
