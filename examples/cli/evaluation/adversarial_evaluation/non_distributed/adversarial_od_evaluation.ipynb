# Non-Distributed Adversarial Evaluation for Object Detection

Adversarial Evaluation is a key component of the robustness evaluation of machine learning models. Such evaluation might be time consuming by its nature. In this notebook, we will demonstrate how to use non-distributed adversarial evaluation for object detection in `advsecurenet` CLI. Distributed adversarial evaluation for object detection is not yet supported.

## Using a Configuration File


### Step 1: Create a Configuration File

As a first step, you need to have a default configuration file. You can create a configuration file by running the following command:
    
```sh
advsecurenet utils configs get -c object_detection_evaluation_config.yml -s -o ./object_detection_evaluation_config.yml
```

Here, we are also providing the `-o` flag to specify the output directory where the configuration file will be saved. You can also specify the output directory by passing the `-o` flag followed by the path to the directory where you want to save the configuration file. If you don't provide the `-o` flag, the configuration file will be saved in the current working directory with the name `object_detection_evaluation_config.yml`. If the file already exists, it will be overwritten. If the output path is a directory, the configuration file will be saved in that directory with the name `object_detection_evaluation_config.yml`.
