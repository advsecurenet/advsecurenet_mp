# Non-Distributed Adversarial Training - Object Detection

`advsecurenet` library provides a simple and easy-to-use API for adversarial training. In this notebook, we will demonstrate how to use the library to adversarially train a simple neural network on the COCO dataset. We will use the non-distributed version of the library in this notebook. As for the current state of the package, it is not possible to run distributed adversarial training for object detection attacks.

## Using a Configuration File

### Step 1: Create a Configuration File

As a first step, you need to have a default configuration file. You can create a configuration file by running the following command:
    
```sh
advsecurenet utils configs get -c adversarial_od_training_config.yml -s -o ./default_advsersarial_od_training_config.yml
```

Here, we are also providing the `-o` flag to specify the output directory where the configuration file will be saved. You can also specify the output directory by passing the `-o` flag followed by the path to the directory where you want to save the configuration file. If you don't provide the `-o` flag, the configuration file will be saved in the current working directory with the name `adversarial_od_training_config.yml`. If the file already exists, it will be overwritten. If the output path is a directory, the configuration file will be saved in that directory with the name `adversarial_od_training_config.yml`.

### Step 2: Modify the Configuration File

The default configuration file is capable of running the adversarial training on the `COCO` dataset using the `CustomYolov5Model` model and `TOG` attack (in order to run the `TOG` attack, you should move the config file inside this folder - similarly as it is done for `DPatch`).

You can modify the configuration file to run the training on a different dataset, model, or change other hyperparameters. You can also add new hyperparameters to the configuration file.

For the sake of this example, we will modify the configuration file to run the training using the `FasterRCNN` model and the `DPatch` attack. We will also decrease the `random_sample_size` to *8* and the `batch_size` to *2*. You can find the updated config file [here](adversarial_od_training_config.yml).






### Step 3: Train the Model

Now, you can adversarially train the model by running the following command:

```sh
advsecurenet defense adversarial-training -c ./adversarial_od_training_config.yml
```
