# Non-Distributed Differential Privacy Training

The `advsecurenet` library provides privacy-preserving model training using differential privacy techniques. In this notebook, we will train a model with differential privacy guarantees in a non-distributed manner using a single GPU or CPU.

## Using a Configuration File

### Step 1: Create a Configuration File

To begin differential privacy training, you need a configuration file. You can create a differential privacy configuration file by running the following command:

```sh
advsecurenet utils configs get -c differential_privacy_train_config.yml -s -o ./differential_privacy_non_distributed.yml
```

The `-o` flag specifies the output directory where the configuration file will be saved. If you don't provide the `-o` flag, the configuration file will be saved in the current working directory. If the file already exists, it will be overwritten.

````python
!advsecurenet utils configs get -c differential_privacy_train_config.yml -s -o ./differential_privacy_non_distributed.yml
````

````python
# Let's check the content of the default config file
!advsecurenet utils configs get -c differential_privacy_train_config.yml -p
````

### Step 2: Review and Modify the Configuration File

The default differential privacy configuration file includes privacy parameters such as noise multiplier, gradient clipping, and delta values. You can modify the configuration file to adjust the privacy-utility trade-off, change the dataset, model architecture, or other training hyperparameters.

Key differential privacy parameters to consider:
- `noise_multiplier`: Controls the amount of noise added during training
- `max_grad_norm`: Maximum L2 norm for gradient clipping  
- `delta`: Privacy parameter that bounds the probability of privacy failure

### Step 3: Train the Model with Differential Privacy

Now you can train the model with differential privacy guarantees by running the following command:

```sh
advsecurenet train -c differential_privacy_non_distributed.yml
```

The training process will apply privacy-preserving techniques including gradient clipping and noise addition to ensure individual training examples remain private while still learning useful patterns.

````python
!advsecurenet train -c differential_privacy_non_distributed.yml
````
