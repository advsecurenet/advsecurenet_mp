# Fine-tuning GPT-2 for Sequence Classification

This notebook demonstrates how to fine-tune GPT-2 for sequence classification using the `ModelLoader`, `EnhancedDataLoader`, and `TrainerPipeline` from the `llm_finetuning` module.

```python
# Setup and imports
import sys
import warnings
warnings.filterwarnings('ignore')

# Add project path and import our enhanced system
sys.path.append('/Users/philip/Desktop/advsecurenet_mp')
from advsecurenet.llm_finetuning import (
    ModelLoader, DataConfig, DataLoader, Config, TrainerPipeline
)

# Load dataset from HuggingFace
from datasets import load_dataset
dataset = load_dataset("ag_news")

print("Dataset loaded successfully!")
```

## Model Loading

```python
# Configure and load GPT-2 model for sequence classification
from transformers import GPT2ForSequenceClassification

# Create a custom GPT-2 model for classification
custom_model = GPT2ForSequenceClassification.from_pretrained("gpt2", num_labels=4)

# Define the configuration
config = Config(
    model_name="gpt2",
    model_obj=custom_model,
    num_labels=4,  # Number of labels in the dataset
    output_dir="./results/gpt2_ag_news"
)

print("Model loaded successfully!")
```

## Data Loading

```python
# Configure data loading
data_config = DataConfig(
    text_column="text",
    label_column="label",
    validation_split=0.2,
    max_length=128
)

# Initialize DataLoader
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
data_loader = DataLoader(tokenizer, data_config)

# Prepare datasets
train_dataset, val_dataset = data_loader.prepare_datasets_from_texts(
    train_texts=dataset["train"]["text"],
    val_texts=dataset["test"]["text"],
    task_type="classification"
)

print("Data loaded and processed successfully!")
```

## Training

```python
# Initialize and run training pipeline
pipeline = TrainerPipeline(
    config=config,
    dataset_input=dataset,
    text_column="text",
    label_column="label",
    split_keys=("train", "test")
)

print("Starting training...")
pipeline.run()

print("Training completed!")
```

Model successfully fine-tuned for sequence classification using the `llm_finetuning` module.
